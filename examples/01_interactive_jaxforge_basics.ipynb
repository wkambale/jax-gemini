# Introduction to Jax Forge
This notebook demonstrates the basics of `jaxforge` (Jax Forge): previewing, building, and refining JAX and Flax NNX models from natural language prompts.

```python
import os

import jaxforge as jf

jf.config.set({
    "model_name": "gemini-3.1-pro",
    "api_key": os.environ.get("GEMINI_API_KEY", "YOUR_API_KEY_HERE"),
})
```

## 1. Previewing and Building a Model
Preview the generated Flax NNX source code and AST validation status first, then execute the plan.

```python
plan = jf.preview(
    "Build a 3-layer MLP classifier with 128 hidden units and ReLU activations"
)
print(plan.code)
model = jf.execute_plan(plan)
print("\nExplanation:", jf.explain())
```

## 2. Interactive Refinement
Refine the existing architecture with follow-up natural language instructions and unified diffs.

```python
model = jf.modify("Add a dropout layer with rate 0.3 before the final layer.")
print(jf.show_code())
print("\nExplanation:", jf.explain())
```
