# Introduction to Jax-Gemini
This notebook demonstrates the basics of jax-gemini: building models out of natural language prompts.

```python
import os

import jax_gemini as jg

jg.config.set({"gemini_api_key": os.environ.get("GEMINI_API_KEY", "YOUR_API_KEY_HERE")})
```

## 1. Building a Model
We can build a model entirely from text.

```python
model = jg.build("Build a 3-layer MLP classifier with 128 hidden units and ReLU activations")
print(jg.show_code())
print("\nExplanation:", jg.explain())
```

## 2. Interactive Refinement
We can modify the model by giving it follow-up instructions.

```python
model = jg.modify("Add a dropout layer with rate 0.3 before the final layer.")
print(jg.show_code())
print("\nExplanation:", jg.explain())
```
