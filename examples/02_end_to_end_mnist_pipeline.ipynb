# End-to-End MNIST Training Pipeline

```python
import os

import numpy as np

import jax_gemini as jg

jg.config.set({"gemini_api_key": os.environ.get("GEMINI_API_KEY", "YOUR_API_KEY_HERE")})
```

```python
# Fake MNIST-like data for demonstration
X_train = np.random.randn(100, 28, 28, 1).astype(np.float32)
y_train = np.random.randint(0, 10, size=(100,))
X_test = np.random.randn(20, 28, 28, 1).astype(np.float32)
y_test = np.random.randint(0, 10, size=(20,))
```

```python
model = jg.build("Build a small CNN for 28x28 grayscale images with 10 output classes")
print(jg.explain())
```

```python
model, metrics = jg.train(
    "Train for 3 epochs using Adam optimizer and cross-entropy loss", dataset=(X_train, y_train)
)
print("Training metrics:", metrics)
```

```python
eval_metrics = jg.evaluate("Evaluate the model returning accuracy", dataset=(X_test, y_test))
print("Evaluation metrics:", eval_metrics)
```

```python
path = jg.save("mnist_cnn_checkpoint")
print(f"Model saved to {path}")
```
